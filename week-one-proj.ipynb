```python
import random

QUESTIONS = [
    {
        "question": "Capital city of Kenya is",
        "choices": {"A": "Nairobi", "B": "Narok", "C": "Nakuru", "D": "Kenya"},
        "answer": "A"
    },
    {
        "question": "What does len([1, 2, 3]) return?",
        "choices": {"A": "2", "B": "3", "C": "4", "D": "Error"},
        "answer": "B"
    },
    {
        "question": "Which data type stores True or False?",
        "choices": {"A": "int", "B": "str", "C": "bool", "D": "list"},
        "answer": "C"
    },
    {
        "question": "Which loop repeats when a condition is true?",
        "choices": {"A": "for", "B": "while", "C": "repeat", "D": "until"},
        "answer": "B"
    },
    {
        "question": "What symbol starts a single-line comment in Python?",
        "choices": {"A": "//", "B": "--", "C": "#", "D": "/*"},
        "answer": "C"
    },
    {
        "question": "Which method adds an item to the end of a list?",
        "choices": {"A": "add()", "B": "insert()", "C": "append()", "D": "push()"},
        "answer": "C"
    },
    {
        "question": "What is the output of 6 / 2?",
        "choices": {"A": "1", "B": "2", "C": "3", "D": "0"},
        "answer": "C"
    },
    {
        "question": "Which function converts a string to an integer?",
        "choices": {"A": "str()", "B": "int()", "C": "float()", "D": "bool()"},
        "answer": "B"
    }
]


def get_answer():
    while True:
        raw = input("Your answer (A, B, C, or D): ").strip().upper()
        if raw in QUESTIONS[0]["choices"]:
            return raw
        print("Invalid input. Please enter A, B, C, or D.")


def feedback(score, total):
    percentage = (score / total) * 100

    if percentage >= 80:
        return "Excellent!"
    elif percentage >= 50:
        return "Good job!"
    return "Try Again."


def run_quiz():
    questions = QUESTIONS.copy()
    random.shuffle(questions)
    score = 0

    print("Welcome to the Python Quiz!")
    print("Answer each question with A, B, C, or D.\n")

    for index, item in enumerate(questions, start=1):
        print(f"Question {index} of {len(questions)}")
        print(item["question"])

        for letter in ["A", "B", "C", "D"]:
            print(f"  {letter}. {item['choices'][letter]}")

        try:
            user_answer = get_answer()
        except (KeyboardInterrupt, EOFError):
            print("\nQuiz ended early.")
            return

        if user_answer == item["answer"]:
            score += 1
            print("Correct!\n")
        else:
            print(f"Incorrect. The correct answer was {item['answer']}.\n")

    total = len(questions)
    print(f"Quiz complete! Your score: {score}/{total}")
    print(f"Percentage: {(score / total) * 100:.1f}%")
    print(feedback(score, total))


if __name__ == "__main__":
    run_quiz()
```
